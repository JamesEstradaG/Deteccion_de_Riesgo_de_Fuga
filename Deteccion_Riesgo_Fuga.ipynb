```python
import pandas as pd
import numpy as np

datos_sucios = {
    'ID_Empleado' : ['101','102','103','104','105'],
    'Nombre_Completo' : ['ana SILVA', 'luis perez', 'MARIA GOMEZ', 'jorge diaz', 'sofia ruiz'],
    'Departamento' : ['Operaciones', 'Ventas', 'RRHH', 'Ventas', 'Operaciones'],
    'Salario_Base' : ['$15,000.00', '12000 MXN', '18000', 'NaN', '$16,000'],
    'Desempenio' : ['Alto', 'Bajo', 'Alto', 'Medio', np.nan]
}

df_empleados = pd.DataFrame(datos_sucios)
print(df_empleados)
```

```python
df_empleados['Nombre_Completo'] = df_empleados['Nombre_Completo'].str.title()

print(df_empleados)
```

```python
df_empleados['ID_Empleado'] = df_empleados['ID_Empleado'].str.strip()

print(df_empleados)
```

```python
df_empleados['Salario_Base'] = df_empleados['Salario_Base'].replace(r'[^\d.]', '', regex=True)

print(df_empleados)
```

```python
df_empleados['Salario_Base'] = pd.to_numeric(df_empleados['Salario_Base'], errors='coerce')

print(df_empleados)
```

```python
promedio_salarial = df_empleados['Salario_Base'].mean()
df_empleados['Salario_Base'] = df_empleados['Salario_Base'].fillna(promedio_salarial)

print(df_empleados)
```

```python
df_empleados['Desempenio'] = df_empleados['Desempenio'].fillna('No Evaluado')

print(df_empleados)
```
